# Clickstream Demo

We demo 2 different functions for collecting clickstream data from Wikipedia:
1. `get_clickstreams` - for collecting different types of clickstream from a list of pages.
2. `pipeline_get_clickstreams` - A convenience wrapper function that, in addition to the above, also sets up session and redirect maps.

## Setup

```python
import wikitoolkit
import pandas as pd

my_agent = 'mwapi testing <p.gildersleve@exeter.ac.uk>'
wtsession = wikitoolkit.WTSession('en.wikipedia', user_agent=my_agent)
toparts = pd.read_csv('data/topviews-2024_07_31.csv')
artlist = toparts['Page'].unique().tolist() # ~1000 top articles yesterday
pagemaps = wikitoolkit.PageMaps()

```

```python
cs = await wikitoolkit.get_clickstreams(wtsession, titles=artlist[:10], pagemaps=pagemaps,
                           date='2022-05-12', consolidate_redirects=True)
cs
```

```python
cs = await wikitoolkit.get_clickstreams(wtsession, titles=artlist[:10], pagemaps=pagemaps,
                           date='2022-05-12', consolidate_redirects=False)
pd.DataFrame(cs)
```

```python
cs = await wikitoolkit.get_clickstreams(wtsession, titles=artlist[:10], pagemaps=pagemaps,
                           daterange=('2022-05-12', '2022-06-13'), consolidate_redirects=True)
pd.DataFrame(cs)
```

```python
cs = await wikitoolkit.pipeline_get_clickstreams('en.wikipedia', my_agent, titles=artlist[:10], pagemaps=pagemaps,
                             gc_args={'date':'2022-05-12'})

pd.DataFrame(cs)
```

```python
import os
datapath = 'data/'
lang = 'en'
monthlist = pd.date_range('2019-12-01', '2024-12-01', freq='MS').strftime('%Y-%m').tolist()
for ym in monthlist:
    urlpath = f'https://dumps.wikimedia.org/other/clickstream/{ym}/clickstream-{lang}wiki-{ym}.tsv.gz'
    if not os.path.exists(datapath):
        os.makedirs(datapath)
    if not os.path.exists(f'{datapath}clickstream-{lang}wiki-{ym}.tsv.gz'):
        print(f'Downloading clickstream data for {ym}...')
        wikitoolkit.download_large_file(urlpath, f'{datapath}clickstream-{lang}wiki-{ym}.tsv.gz')
```
